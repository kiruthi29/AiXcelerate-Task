```python
# DATASET: 8 Happy and 7 Sad sentences

train_data = [
    ("I got full marks in the test today!", "Happy"),
    ("We won the college cricket match", "Happy"),
    ("My friends planned a surprise party for me", "Happy"),
    ("The food in the canteen was really tasty", "Happy"),
    ("I am so excited for the weekend trip", "Happy"),
    ("Our project got selected for the exhibition", "Happy"),
    ("I finally learned how to ride a bicycle", "Happy"),
    ("Mom made my favourite sweets today", "Happy"),
    ("I lost my phone on the bus", "Sad"),
    ("I failed in the mathematics exam", "Sad"),
    ("My best friend is moving to another city", "Sad"),
    ("It rained and our picnic was cancelled", "Sad"),
    ("I missed the last bus and felt very upset", "Sad"),
    ("My laptop stopped working before the submission", "Sad"),
    ("I felt lonely during the long holidays", "Sad"),
]


# ==========================================
# LEVEL 1: COUNT THE MOODS
# ==========================================

def count_moods(data):

    happy_count = 0
    sad_count = 0

    for sentence, mood in data:

        if mood == "Happy":
            happy_count += 1

        elif mood == "Sad":
            sad_count += 1

    return happy_count, sad_count


# ==========================================
# LEVEL 2: GUESS THE MOOD
# Roll number ends in 0, so N = 10
# ==========================================

happy_words = [
    "full",
    "won",
    "surprise",
    "tasty",
    "excited",
    "selected",
    "learned",
    "favourite",
    "marks",
    "bicycle"
]

sad_words = [
    "lost",
    "failed",
    "moving",
    "cancelled",
    "missed",
    "upset",
    "stopped",
    "lonely",
    "rained",
    "working"
]


# Function to remove punctuation and lowercase words

def clean_words(sentence):

    cleaned_sentence = ""

    for character in sentence.lower():

        if character.isalnum() or character.isspace():
            cleaned_sentence += character

        else:
            cleaned_sentence += " "

    return cleaned_sentence.split()


def guess_mood(sentence):

    words = clean_words(sentence)

    happy_count = 0
    sad_count = 0

    for word in words:

        if word in happy_words:
            happy_count += 1

        if word in sad_words:
            sad_count += 1

    if happy_count > sad_count:
        return "Happy"

    elif sad_count > happy_count:
        return "Sad"

    # Tie-breaking rule: Choose Sad as the default
    # because no clear positive clue means we avoid assuming happiness.

    else:
        return "Sad"


# ==========================================
# LEVEL 3: TEST YOUR PROGRAM
# ==========================================

practice_data = [
    ("I won a medal in the sports meet", "Happy"),
    ("I lost my notebook before the exam", "Sad"),
    ("My friends gave me a lovely surprise", "Happy"),
    ("The train was late and I felt upset", "Sad"),
    ("Today was a great day", "Happy"),
]


def calculate_accuracy(data, predictor):

    correct = 0

    for sentence, actual_mood in data:

        predicted_mood = predictor(sentence)

        if predicted_mood == actual_mood:
            correct += 1

    accuracy = (correct / len(data)) * 100

    return accuracy


# ==========================================
# BONUS: LEARN WORD SCORES AUTOMATICALLY
# ==========================================

def learn_scores(data):

    scores = {}

    for sentence, mood in data:

        words = clean_words(sentence)

        for word in words:

            if word not in scores:
                scores[word] = 0

            if mood == "Happy":
                scores[word] += 1

            elif mood == "Sad":
                scores[word] -= 1

    return scores


# Train the word-score model using training data only

learned_scores = learn_scores(train_data)


# ==========================================
# BONUS: PREDICT USING LEARNED SCORES
# ==========================================

def guess_mood_learned(sentence):

    words = clean_words(sentence)

    total_score = 0

    for word in words:

        total_score += learned_scores.get(word, 0)

    if total_score > 0:
        return "Happy"

    else:
        return "Sad"


# ==========================================
# MAIN PROGRAM
# ==========================================

if __name__ == "__main__":

    # LEVEL 1

    print("=" * 45)
    print("LEVEL 1: COUNT THE MOODS")
    print("=" * 45)

    print("Mood counts:", count_moods(train_data))


    # LEVEL 2

    print("\n" + "=" * 45)
    print("LEVEL 2: RULE-BASED MOOD PREDICTION")
    print("=" * 45)

    for sentence, actual in practice_data:

        prediction = guess_mood(sentence)

        print("\nSentence:", sentence)
        print("Predicted:", prediction)
        print("Actual:", actual)


    level2_accuracy = calculate_accuracy(
        practice_data, guess_mood
    )

    print("\nLevel 2 Accuracy:", level2_accuracy, "%")


    # BONUS

    print("\n" + "=" * 45)
    print("BONUS: LEARNED WORD-SCORE MODEL")
    print("=" * 45)

    for sentence, actual in practice_data:

        prediction = guess_mood_learned(sentence)

        print("\nSentence:", sentence)
        print("Predicted:", prediction)
        print("Actual:", actual)


    bonus_accuracy = calculate_accuracy(
        practice_data, guess_mood_learned
    )

    print("\nBonus Model Accuracy:", bonus_accuracy, "%")


    # COMPARISON

    print("\n" + "=" * 45)
    print("MODEL COMPARISON")
    print("=" * 45)

    print("Rule-Based Accuracy:", level2_accuracy, "%")
    print("Learned Model Accuracy:", bonus_accuracy, "%")
```
